# Joint UMAP — Topic Scatter Maps

Topic maps on the **shared** joint-UMAP canvas (loaded from `03-compute_coords`).
Because every map uses the same axis limits and an equal aspect ratio, the
papers, teams, and overlay maps are geometrically comparable — a teams-only map
shows teams in their real sub-region instead of being stretched to fill the frame.

Each map can be coloured by hierarchy level and labelled in two styles:

- **levels** — `micro` (low-level clusters), `meso` (mid groups), `macro` (high groups)
- **label styles** — `side` (gutter labels with connectors) or `overlay` (labels over clusters)

> Run `03-compute_coords.ipynb` first.

```python
# ── Make the shared aux/ package importable ───────────────────────────────────
# These notebooks live in 05-reporting/ alongside the aux/ package and
# setup_run.py.
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd()))
```

```python
import matplotlib.pyplot as plt

from aux.paths import set_seed
from setup_run import setup
from aux.coords import load_plot_data
from aux.scatter import plot_topic_map, plot_overlay

set_seed()
RUN = setup()  # today's run folder: assets/<date>/05/
df_papers, df_teams, XLIM, YLIM = load_plot_data(RUN)
print(f"Papers: {len(df_papers):,} | Teams: {len(df_teams):,}")
print(f"Shared canvas: x={tuple(round(v, 2) for v in XLIM)}, y={tuple(round(v, 2) for v in YLIM)}")

LEVELS = ["micro", "meso", "macro"]   # low / mid / high hierarchy
STYLES = ["side", "overlay"]
```

## Papers — every level × label style

```python
for level in LEVELS:
    for style in STYLES:
        fig, _ = plot_topic_map(
            df_papers, level=level, label_style=style, xlim=XLIM, ylim=YLIM,
            title=f"Papers — {level} clusters ({style} labels)",
            size_col="cited_by_count", min_docs=20,
            out_path=RUN.out(f"umap_papers_{level}_{style}.png"),
        )
        plt.show()
```

## Teams — every level × label style

```python
for level in LEVELS:
    for style in STYLES:
        fig, _ = plot_topic_map(
            df_teams, level=level, label_style=style, xlim=XLIM, ylim=YLIM,
            title=f"Teams — {level} clusters ({style} labels)",
            min_docs=10,
            out_path=RUN.out(f"umap_teams_{level}_{style}.png"),
        )
        plt.show()
```

## Overlay — papers (muted) + teams (coloured), per level

```python
for level in LEVELS:
    fig, _ = plot_overlay(
        df_papers, df_teams, XLIM, YLIM, level=level,
        title=f"Overlay — Papers (muted) + Teams ({level} clusters)",
        out_path=RUN.out(f"umap_overlay_{level}.png"),
    )
    plt.show()
```
