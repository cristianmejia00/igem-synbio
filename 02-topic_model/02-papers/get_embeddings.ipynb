# Compute Sentence-Transformer Embeddings — SynBio OpenAlex Papers

Reads the **SynBio OpenAlex Papers** dataset, concatenates title + abstract for each
record, computes embeddings with a sentence-transformer model, and saves the
results (embedding matrix + aligned corpus) to `assets/embeddings/`.

```python
# ── Make the shared aux/ package importable ───────────────────────────────────
# These notebooks live one level below 02-topic_model/, where the aux/ package
# resides; add that folder to the import path.
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
```

```python
import pandas as pd

from aux.paths import ASSETS_DIR, EMBEDDINGS_DIR
from aux.embeddings import prepare_text, encode_texts, save_embeddings

# ── CONFIG: SynBio OpenAlex Papers ──────────────────────────────────────────────────
RAW_FILE        = "synbio_openalex.txt"          # source dataset in assets/
TITLE_COL       = "title"
ABSTRACT_COL    = "abstract"
ID_COL          = "id"
EMBEDDINGS_FILE = "papers_embeddings.npy"
CORPUS_FILE     = "papers_corpus.txt"
```

## 1. Load dataset

```python
papers = pd.read_csv(ASSETS_DIR / RAW_FILE, sep="\t")
print(f"SynBio OpenAlex Papers: {len(papers):,} rows")
```

## 2. Prepare text for embedding

Concatenate title and abstract into a single `text` field, clean minimally
(keep alphabetic characters, collapse whitespace), and drop rows with no text.

```python
corpus = prepare_text(papers, title_col=TITLE_COL, abstract_col=ABSTRACT_COL)
print(f"Corpus: {len(corpus):,} docs")
```

## 3. Compute embeddings

```python
embeddings = encode_texts(corpus["text"].tolist())
print(f"shape: {embeddings.shape}")
```

## 4. Save embeddings and corpus metadata

```python
save_embeddings(embeddings, corpus, ID_COL, EMBEDDINGS_FILE, CORPUS_FILE)
print(f"Saved {len(corpus):,} docs → {EMBEDDINGS_DIR}")
print(f"  {EMBEDDINGS_FILE}")
print(f"  {CORPUS_FILE}")
```
