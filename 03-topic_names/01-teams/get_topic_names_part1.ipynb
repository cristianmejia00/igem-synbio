# Topic Naming with OpenAI — iGEM Teams (Part 1)

Loads the **iGEM Teams** topic info and corpus produced in
`02-topic_model/`, then sends representative documents per cluster to GPT to
obtain:

1. A **description** of each topic cluster.
2. An **enhanced (synthesised) description**.
3. A short **cluster name**.

Prompts are loaded from `prompts.yaml`. Results are saved as
`teams_topic_names.txt` alongside the topic models in
`assets/topic_models/`.

```python
# ── Make the shared aux/ package importable ───────────────────────────────────
# These notebooks live one level below 03-topic_names/, where the aux/ package
# resides; add that folder to the import path.
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
```

```python
from aux.paths import MODELS_DIR, OPENAI_MODEL, TOP_N_DOCS
from aux.openai_client import load_prompts, make_client
from aux.tables import load_topic_corpus, save_topic_names
from aux.naming import name_topics

# ── CONFIG: iGEM Teams ──────────────────────────────────────────────────
PREFIX = "teams"
ID_COL = "UT"
MODEL  = OPENAI_MODEL          # OpenAI model used for naming

prompts = load_prompts()
client = make_client()
print(f"Theme: {prompts['theme']}")
```

## 1. Load topic info and corpus

```python
topic_info, df = load_topic_corpus(PREFIX, ID_COL)
n_topics = topic_info.Topic.max() + 1
print(f"Teams: {len(df):,} docs, {n_topics} topics")
```

## 2. Generate descriptions and names

```python
print("Naming iGEM Teams topics...")
names = name_topics(df, topic_info, client, prompts, model=MODEL, top_n=TOP_N_DOCS)
names
```

## 3. Save results

```python
save_topic_names(names, PREFIX)
print(f"Saved {len(names)} topic names → {MODELS_DIR / f'{PREFIX}_topic_names.txt'}")
```
